```python
#Part 2

#fill embarked with the mode
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

#strategy 1
df['mean_age_filled'] = df['Age'].fillna(df['Age'].mean())

#strategy 2
df['median_age_filled'] = df.groupby(['Sex', 'Pclass'])['Age'].transform(lambda x: x.fillna(x.median()))
```

```python
#Part 3

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

print(f"Minimum FamilySize: {df['FamilySize'].min()}")
print(f"Maximum FamilySize: {df['FamilySize'].max()}")
print(f"Mean FamilySize: {df['FamilySize'].mean():.2f}")

count_alone = (df['FamilySize'] == 1).sum()
print(f"Passengers traveling alone: {count_alone}")
```

```python
#Part 4

df_female = df[df['Sex'] == 'female']
df_male = df[df['Sex'] == 'male']
df_under_16 = df[df['Age'] < 16]

print(f"Survival Rate (female): {df_female['Survived'].mean()}")
print(f"Survival Rate (male): {df_male['Survived'].mean()}")
print(f"Survival Rate (under 16): {df_under_16['Survived'].mean()}")

grouped = df.groupby(['Sex', 'Pclass'])['Survived'].mean()
print("\n Grouped Survival Rates ")
print(grouped)
```

```python
#Part 5

count_tickets_once = df.drop_duplicates(subset=['Ticket'])

average_fare_by_pclass = count_tickets_once.groupby('Pclass')['Fare'].mean()

print("Average Fare for each Pclass (counting shared tickets too):")
print("Pclass 1: $", round(average_fare_by_pclass[1], 2))
print("Pclass 2: $", round(average_fare_by_pclass[2], 2))
print("Pclass 3: $", round(average_fare_by_pclass[3], 2))
```
